## Steps of a Machine Learning project

- Look at the big picture.
- Get the data.
- Discover and visualize the data to gain insights.
- Prepare the data for Machine Learning algorithms.
- Select a model and train it.
- Fine-tune your model.
- Present your solution.
- Launch, monitor, and maintain your system.

**1. Frame the problem**
   - What is exactly the business objective
   - How does the company expect to use and benefit from this model?
   - What performance measure you will use to evaluate your model

**PIPELINES**

A sequence of data processing components is called a data pipeline.

Components typically run asynchronously.

Each component pulls in a large amount of data, process it and spits out the result in another data store and then this process continues for the other components in the pipeline.

**Univariate Regression:** A regression model that uses only one feature to estimate the target feature.

**Multivariate Regression:** A regression model will use multiple features to predict the target feature.

**2. Select a Performance Measure**

A typical measure is the `Root Mean Square(RMSE)` which measures the *standard deviation* of the errors the system makes in it's predictions.
